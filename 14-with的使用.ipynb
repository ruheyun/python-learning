with 关键字是用来处理 python 中的资源管理问题的，通过 with 可以自动释放资源。

```python
file = open('example/example.txt', 'r')
try:
    content = file.read()
finally:
    file.close()
print(content)
```

通过 open 打开资源后，需要在后面使用 close 对资源进行释放，但可能会漏写 finally 分支部分。

```python
with open('example/example.txt', 'r') as file:
    content = file.read()
    print(content)
```

通过 with 会自动释放资源，且代码简洁。
