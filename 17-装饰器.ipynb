装饰器是 python 中的高级功能，用于在不修改原函数的前提下，动态扩展函数或类的功能。

本质上，装饰器是一个高级函数，接收函数作为参数，然后返回一个新函数。

通常，使用方法是在函数的前面加上 @decorator_name

```python
def decorator_fun(original_fun):
    def wrapper(*args, **kwargs):
        print('打印前')
        result = original_fun(*args, **kwargs)
        print('打印后')

        return result
    
    return wrapper

@decorator_fun
def target_fun():
    print('原函数执行')

target_fun()
```

上述等价写法：target_fun = decorator_fun(target_fun)

调用 target_fun 相当于调用 wrapper

@ 在 python 称为语法糖，装饰器就是在原函数前面使用语法糖，本质是函数替换

```python
# 原函数中有参数的时候，wrapper 需要使用 args 和 kwargs 接收
def decorator_fun(original_fun):
    def wrapper(*args, **kwargs):
        print('before print')

        original_fun(*args, **kwargs)

        print('after print')

        return 

    return wrapper

@decorator_fun
def target_fun(name):
    print(f'your name: {name}')

target_fun('ruhe')
```

带参数的装饰器

```python
# 如果装饰器带参数，则需要再装饰器函数外加一个外层函数

def repeat(num_times):
    def decorator(func):
        def wrapper(*args, **kwargs):
            for _ in range(num_times):
                func(*args, **kwargs)
        return wrapper

    return decorator

@repeat(3)
def target():
    print('hello!')

target()
```

函数形式的类装饰器

```python
def decorator(cls):
    class Wrapper:
        def __init__(self, *args, **kwargs):
            self.wrapped = cls(*args, **kwargs)

        def __getattr__(self, name):
            return getattr(self.wrapped, name)

        def display(self):
            print('打印前')
            self.wrapped.display()
            print('打印后')

    return Wrapper

@decorator
class MyClass:
    def display(self):
        print('原方法')

    def print_hello(self):
        print('hello')

mc = MyClass()
mc.display()

# 当装饰器里没有原类的方法，会通过 __getatter__ 直接调用原类的方法
mc.print_hello()
```

类形式的类装饰器

```python
class SingletonDecorator:
    def __init__(self, cls):
        self.cls = cls
        self.instance = None
    def __call__(self, *args, **kwds):
        if self.instance is None:
            self.instance = self.cls(*args, **kwds)
        return self.instance

@SingletonDecorator
class DataBase:
    def __init__(self):
        print('初始化')

db1 = DataBase()
db2 = DataBase()

print(db1 is db2)
    
```

多个装饰器堆叠，从上往下执行

```python
def decorator1(func):
    def wrapper():
        print('decorator1')
        func()
    return wrapper

def decorator2(func):
    def wrapper():
        print('decorator2')
        func()

    return wrapper

@decorator1
@decorator2
def target():
    print('原函数')

target()
```
