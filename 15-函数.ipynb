函数是组织好的，可重复使用的代码段。

如下是一个简单的函数及其调用例子，使用 def 关键字开头

```python
def print_hello():
    print('hello!')

print_hello()
```

在函数名后的小括号里可以带上参数

```python
def print_message(message):
    print(message)

print_message('hello world!')
```

```python
# 可传递的参数包括所有的 python 数据类型
print_message(['hello', 12])
```

需要注意的是：对于可变类型数据，参数传递的是‘地址’；对于不可变类型数据，参数传递的是值。
因此对于可变类型数据，在函数内进行更改时，会同时改变函数外的参数。

```python
def print_info(a, b):
    a = 10
    b.append(10)

a = 1
b = [1, 2]
print(a, id(a))
print(b, id(b))

print_info(a, b)

print(a, id(a))
print(b, id(b))
```

上述中 a 是整数类型数据，不可变，所以值传递，在函数内部更改不会影响外部参数；

b 是列表类型数据，可变数据，所以引用传递，在函数内部更改会影响外部参数。

在函数中共有四种参数：

必须参数

关键字参数

默认参数

不定长参数

```python
# 必须参数，如果不传递参数会报错

def fun(a):
    print(a)

fun()
```

```python
# 关键字参数，通过关键字传参可以不按照顺序

def fun(a, b):
    print(a, b)

fun(b=2, a=1)
```

```python
# 默认参数，如果设置了默认参数，则函数调用时可以不传参

def fun(a=1):
    print(a)

fun()
```

```python
# 不定长参数，通过 * 号可以接收任意多个参数，会保存到一个元组中，不能使用关键字传参

def fun(*params):
    print(params)

fun()

fun(1)

fun(1, 2, 3)
```

```python
# 不定长参数，通过 ** 两个星号，会将参数保存为字典，而且传参时必须用关键字传参

def fun(**params):
    print(params)

fun(a=1, b=2)
```

强制位置参数，通过 / 和 * 来进行指定，/ 前参数只能使用形参传参，* 后面参数只能使用关键字传参，两者之间参数都行，不作强制要求

```python
def fun(a, b , /, c, d, *, e, f):
    print(a, b, c, d, e, f)

fun(1, 2, 3, d=4, e=5, f=6)
```

匿名函数：使用 lambda 关键字来定义

```python
x = lambda x: x + 10

x(5)
```

退出函数，可以使用关键字 return，return 会直接退出函数，并返回其后面的数据内容

```python
def fun(a):
    a += 10
    return a

a = fun(a)
a
```
