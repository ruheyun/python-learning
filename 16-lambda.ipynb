lambda 函数是匿名的，它们没有函数名称，只能通过赋值给变量或作为参数传递给其他函数来使用。

主要作用于map()、filter()、reduce() 等函数中。

1. 作用于 map 中

```python
nums = [1, 2, 3, 0, 6, 5]

squrd = list(map(lambda x: x * x, nums))

squrd
```

2. 作用于 filter 中

```python
even = list(filter(lambda x: x % 2 == 0, nums))
even
```

3. 作用于 reduce 中

```python
from functools import reduce
prod = reduce(lambda x, y: x * y, nums)
prod
```

```python
prod = reduce(lambda x, y: x + y, nums)
prod
```
